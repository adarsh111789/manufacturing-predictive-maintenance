```python
spark.conf.set(
    "***"
)
```

```python
raw_path = "abfss://raw-data@adarshpmstorage01.dfs.core.windows.net/*.csv"

bronze_df = spark.read.csv(
    raw_path,
    header=True,
    inferSchema=True
)

display(bronze_df)
```

```python
sanitized_columns = [
    col_name.replace("(", "_")
            .replace(")", "")
            .replace("?", "")
            .replace(" ", "_")
            .replace("-", "_")
    for col_name in bronze_df.columns
]

bronze_df = bronze_df.toDF(*sanitized_columns)

bronze_df.printSchema()
```

```python
bronze_path = "abfss://processed-data@adarshpmstorage01.dfs.core.windows.net/bronze"

bronze_df.write.format("delta").mode("append").save(bronze_path)
```

```python
verify_bronze = spark.read.format("delta").load(bronze_path)

verify_bronze.printSchema()
display(verify_bronze)
```
